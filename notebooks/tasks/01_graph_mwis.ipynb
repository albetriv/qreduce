# Grafo di riduzione e MWIS

Esempio sintetico: la certificazione riguarda solo questa piccola istanza MWIS, non il problema originale.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

from qreduce.notebooks.examples import graph_example
from qreduce.notebooks.plots import plot_reduction_graph

example = graph_example()
```

## Dati di input: nodi, archi e pesi

I pesi sono assegnati ai nodi nella definizione del problema, prima della risoluzione MWIS.


```python
display(pd.DataFrame(example["nodes"])[["node", "weight"]])
display(pd.DataFrame(example["edges"]))
```

```python
pd.DataFrame(example["nodes"]).set_index("node")["weight"].plot.bar(title="Pesi dei nodi")
```

## Soluzione esatta diagnostica

La selezione è calcolata a partire dal grafo e dai pesi di input.


## Visualizzazione del grafo

I nodi verdi appartengono alla soluzione MWIS esatta. Il layout è illustrativo.

```python

fig, ax = plot_reduction_graph(example["graph"], example["selected_ids"],
                     highlighted_label="MWIS esatta", title="Grafo e MWIS esatta")
plt.show()
```
