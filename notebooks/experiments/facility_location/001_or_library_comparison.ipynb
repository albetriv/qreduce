Riepilogo dei run già eseguiti. Nessun solver viene avviato.

Eseguire il notebook dalla root del repository o da una sottodirectory;
il codice cerca automaticamente `pyproject.toml` per individuare la root.

```python
from pathlib import Path

import pandas as pd

from qreduce.experiments.facility_location.analysis import summary

repository_root = next(
    (directory for directory in (Path.cwd(), *Path.cwd().parents)
     if (directory / 'pyproject.toml').is_file()),
    None,
)
if repository_root is None:
    raise RuntimeError('Run this notebook from inside the QReduce repository.')

output_root = repository_root / 'outputs/facility_location'
run_directories = sorted(path for path in output_root.iterdir() if path.is_dir()) if output_root.exists() else []
results = pd.DataFrame([summary(path) for path in run_directories])
results.reindex(columns=['run_id', 'status', 'objective', 'difference_from_reference', 'fixed_cost', 'allocation_cost', 'open_capacity', 'capacity_utilization', 'reduction_binary_free', 'reduction_continuous_free', 'end_to_end_seconds', 'graph_build_seconds', 'recovery_attempts'])
```
