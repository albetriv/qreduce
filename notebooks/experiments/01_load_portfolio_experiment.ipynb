# Analisi di un esperimento portfolio

Questo notebook carica artifact gia` pubblicati dal runner interno. Non esegue la pipeline QReduce.

```python
from pathlib import Path

import pandas as pd
from IPython.display import display

from qreduce.experiments import load_experiment_run

run_directory = Path("outputs/portfolio-reference/run-001")
artifacts = load_experiment_run(run_directory)
```

```python
display(pd.Series(artifacts.manifest, name="value").to_frame())
display(pd.Series(artifacts.result["run_result"], name="value").to_frame())
display(pd.Series(artifacts.result["metrics"], name="value").to_frame())
```

```python
if artifacts.canonical_mwis_problem is not None:
    graph = artifacts.canonical_mwis_problem["graph"]
    nodes = pd.DataFrame(
        {"item_id": graph["node_ids"], "weight": [graph["node_weights"][node_id] for node_id in graph["node_ids"]]}
    )
    edges = pd.DataFrame(graph["edge_reasons"])
    display(nodes)
    display(edges)
```
